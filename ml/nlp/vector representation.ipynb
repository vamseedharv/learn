```python
import pandas as pd
messages = pd.read_csv('spam.csv', sep=',', encoding='latin-1')
print(messages.head())

```

```python
# data cleaning and preprocessing

import re
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
corpus = []
for i in range(0, len(messages)):
    review = re.sub('[^a-zA-Z]', ' ', messages['message'][i])
    review = review.lower()
    review = review.split()
    review = [ps.stem(word) for word in review if not word in set(stopwords.words('english'))]
    review = ' '.join(review)
    corpus.append(review)

print(corpus)
```

```python
# Create bag of words model
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, binary=True)
X = cv.fit_transform(corpus).toarray()

X
```

## TF-IDF

Term frequency - inverse document frequency
Best than BOG

Advantages - 1. Intuitive 2. Fixed vocab size 3. word importance is being captured(word appears in all sentence is less priorotized and rare occurence is given high priority) 4. ||
Disadvantages - 1. sparsity still exist 2. OOV

```python
import pandas as pd
messages = pd.read_csv('spam.csv', sep=',', encoding='latin-1')
print(messages.head())
```

```python
from nltk.stem import WordNetLemmatizer
wordLemmatize=WordNetLemmatizer()
corpus=[]
for i in range(0, len(messages)):
    review = re.sub('[^a-zA-Z]', ' ', messages['message'][i])
    review = review.lower()
    review = review.split()
    review = [wordLemmatize.lemmatize(word) for word in review if not word in set(stopwords.words('english'))]
    review = ' '.join(review)
    corpus.append(review)

corpus
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(max_features=5000, ngram_range=(1, 2))
tfidf_matrix = tfidf.fit_transform(corpus).toarray()
print(tfidf.vocabulary_) #most occuring features/words combination base don ngram range
print(tfidf_matrix)
```
